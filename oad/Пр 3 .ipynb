```python
df_clean = df.dropna(subset=["Year"])


```

```python
print("Размер очищенного датасета:", df_clean.shape)

```

```python
df_clean.isnull().sum()
```

```python
print("Жанры:", df_clean["Genre"].unique())
print("Количество жанров:", df_clean["Genre"].nunique())

print("\nПлатформы:", df_clean["Platform"].unique())
print("Количество платформ:", df_clean["Platform"].nunique())

print("\nКоличество издателей:", df_clean["Publisher"].nunique())
```

```python
genre_count = df_clean.groupby("Genre")["Name"].count()

genre_count.sort_values(ascending=False)
```

```python
genre_sales = df_clean.groupby("Genre")["Global_Sales"].sum()

genre_sales.sort_values(ascending=False)
```

```python
top10 = df_clean.nlargest(10, "Global_Sales")
top10[["Rank", "Name", "Platform", "Genre", "Global_Sales"]]
```

```python
platform_count = df_clean.groupby("Platform")["Name"].count()

platform_count.sort_values(ascending=False).head(10)
```

```python
platform_sales = df_clean.groupby("Platform")["Global_Sales"].sum()
platform_sales.sort_values(ascending=False).head(10)
```

```python
genre_sales_sorted = genre_sales.sort_values(ascending=False)

plt.figure(figsize=(10, 6))
plt.bar(genre_sales_sorted.index, genre_sales_sorted.values, color="steelblue")
plt.xlabel("Жанр")
plt.ylabel("Мировые продажи, млн")
plt.title("Продажи видеоигр по жанрам")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df_clean[["NA_Sales", "Global_Sales"]].corr()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    df_clean["NA_Sales"],
    df_clean["Global_Sales"],
    alpha=0.5
)
plt.xlabel("Продажи в Северной Америке, млн")
plt.ylabel("Мировые продажи, млн")
plt.title("Связь продаж в Северной Америке и мировых продаж")
plt.tight_layout()
plt.show()
```

```python
df_clean[["NA_Sales", "EU_Sales", "JP_Sales", "Other_Sales", "Global_Sales"]].corr()
```

Сильнее всего связаны продажи в Северной Америке и Европе (коэффициент около 0.77) - западные рынки имеют схожие игровые предпочтения. Продажи в Японии слабее коррелируют с остальными регионами из-за специфики местного рынка. Все региональные продажи сильно связаны с Global_Sales, но NA_Sales показывает самую высокую корреляцию, так как Северная Америка - крупнейший рынок в датасете.
