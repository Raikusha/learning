```python
plt.figure(figsize=(10, 6))
plt.hist(df['popularity'], bins=20, edgecolor='black', color='skyblue')
plt.title('Распределение популярности треков')
plt.xlabel('Популярность')
plt.ylabel('Количество треков')
plt.grid(axis='y', alpha=0.5)
plt.show()
```

Основная часть значений popularity находится в диапазоне от 50 до 85.
Значения ниже 30 и выше 95 встречаются значительно реже остальных.
